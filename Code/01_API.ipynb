### API

#### Diversity Corpus

```python
BASE_URL = "https://www.federalregister.gov/api/v1/documents.json"
KEYWORDS = "diversity"
PER_PAGE = 1000  # max allowed
MAX_PAGES = 20   # adjust as needed
SLEEP = 15        # polite delay to avoid rate limits

diversity_results = []

for page in range(1, MAX_PAGES + 1):
    params = {
        "conditions[term]": KEYWORDS,
        "conditions[publication_date][gte]": "2017-01-20",
        "per_page": PER_PAGE,
        "page": page,
        "order": "relevance"
    }
    print(f"Fetching page {page}...")
    r = requests.get(BASE_URL, params=params)
    data = r.json()
    diversity_results.extend(data.get("results", []))
    
    # stop if there are no more pages
    if not data.get("results"):
        break
    
    time.sleep(SLEEP)
```

```python
for result in diversity_results:
    text = result['excerpts']
    clean_text = re.sub(r'<[^>]+>', '', text)
    result['excerpts'] = re.sub('\n', '', clean_text)
```

```python
filename = 'Raw_Data/diversity_data.json'

with open(filename, 'w') as file:
    json.dump(diversity_results, file, indent=4)
```
